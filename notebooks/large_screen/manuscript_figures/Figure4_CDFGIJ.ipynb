# Figure 4C, D, F, G, I and J — where each factor's genes are expressed

Scores every cell for the outlier gene sets of the fifteen independent
components and draws each score on the UMAP.

Each factor contributes two sets, so thirty maps are produced. Which six of
them are panels C, D, F, G, I and J is not recorded anywhere in the analysis,
so all thirty are kept and the selection is made when the figure is assembled.

## Setup

```python
import scanpy as sc
import pandas as pd

DATA_DIR   = "/home/eraslab1/Projects/E3Ligase/analysisSingle"
ADATA_FILE = f"{DATA_DIR}/outputs/anndata/OriginalFiles/adata-hash-features_singlets_SingleKO_08122020_PerGENE.h5ad"
FACTOR_FILE = f"{DATA_DIR}/TextFiles/ICA_15_factors_effectedGenes_paper.csv"

# Where sc.pl.* writes its PDFs. The original notebook chdir'd into the project
# directory to get the same effect; setting it here avoids the side effect.
sc.settings.figdir = "figures"
```

## The single-knockout object and the factor gene sets

Guides are merged at the target gene level in this object, and it carries the
UMAP the maps are drawn on.

Column names survive the round trip through R as `Factor.1.gene.group.1`; the
two substitutions below turn the gene columns into `Factor1_1` and leave the
guide columns without an underscore, which is what selects them further down.

```python
adataC = sc.read(ADATA_FILE)

cGenes = pd.read_csv(FACTOR_FILE)
cGenes.columns = [x.replace(" ", "") for x in cGenes.columns]
cGenes.columns = [x.replace("genegroup", "_") for x in cGenes.columns]

geneSetColumns = [col for col in cGenes.columns if "_" in col]
print(adataC.shape, "|", len(geneSetColumns), "gene sets")
```

## One score map per gene set

Short columns are padded with `" "` in the file. Those blanks are dropped
before scoring rather than left for `score_genes` to discard, so that the
"not in var_names" warning only fires for genes that are genuinely absent —
a handful whose names did not survive the round trip through R.

```python
varNames = set(adataC.var_names)

for col in geneSetColumns:
    geneList = [str(x).replace(".", "-") for x in cGenes[col]]
    geneList = [g for g in geneList if g.strip() and g != "nan"]   # drop the " " padding

    missing = [g for g in geneList if g not in varNames]
    if missing:
        print(f"{col}: {len(missing)} of {len(geneList)} not in var_names -> {missing}")

    sc.tl.score_genes(adata=adataC, gene_list=geneList, score_name=col)
    sc.pl.umap(adataC, color=col, size=1, color_map="coolwarm", save=col + ".pdf")
```
