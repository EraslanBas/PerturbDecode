```r
source("Main.R")
source("Conf.R")
source("Utilities.R")
library("factoextra")
library("maptree")
library(repr)
```

```r
res = read.csv(file.path("TextFiles", "KOModule_leidenInteractions.csv"), row.names = 1)
```

```r
head(res)
```

```r
res = res[res$term %ni% c("(Intercept)", "n_genes", "mt_frac", paste0("leiden",1:9)),]
```

```r
coefMat = data.frame(reshape(res[,c("term","estimate","respGene")], idvar = "term",
                  timevar = "respGene", direction="wide"))

pValMat = data.frame(reshape(res[,c("term","p.value","respGene")], idvar = "term",
                  timevar = "respGene", direction="wide"))

rownames(coefMat) = coefMat$term
rownames(pValMat) = pValMat$term
coefMat$term = NULL
pValMat$term = NULL

colnames(coefMat) <- sapply(colnames(coefMat), function(x){strsplit(x,"estimate.")[[1]][2]})
colnames(pValMat) <- sapply(colnames(pValMat), function(x){strsplit(x,"p.value.")[[1]][2]})
```

```r
pValsAllAdj <- data.frame(sapply(pValMat, function(x){p.adjust(x, 
                                                  method = "fdr",
                                                  n = length(x))}))
rownames(pValsAllAdj) <- rownames(pValMat)

coef_norm = copy(coefMat)
coef_norm[pValsAllAdj > 0.1] = 0
```

```r
coef_norm_geneModules = coef_norm[,geneModules$GeneName]
```

```r
rownames(coef_norm_geneModules) <- sapply(rownames(coef_norm_geneModules), function(x){gsub("K_0", "M2", x) })
rownames(coef_norm_geneModules) <- sapply(rownames(coef_norm_geneModules), function(x){gsub("K_1", "M3", x) })
rownames(coef_norm_geneModules) <- sapply(rownames(coef_norm_geneModules), function(x){gsub("K_2", "M6", x) })
rownames(coef_norm_geneModules) <- sapply(rownames(coef_norm_geneModules), function(x){gsub("K_3", "M5", x) })
rownames(coef_norm_geneModules) <- sapply(rownames(coef_norm_geneModules), function(x){gsub("K_4", "M1", x) })
rownames(coef_norm_geneModules) <- sapply(rownames(coef_norm_geneModules), function(x){gsub("K_5", "M4", x) })
```

```r
for(elem in seq(1,10,1)){
    rownames(coef_norm_geneModules) <- sapply(rownames(coef_norm_geneModules), function(x){gsub(paste0("leiden",elem),
                                                                                                paste0("C",elem+1), x) })

}
```
