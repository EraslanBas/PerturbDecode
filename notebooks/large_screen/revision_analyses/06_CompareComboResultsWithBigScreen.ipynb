```r
extraGenesAvail = extraGenes[extraGenes %in% colnames(coefsAll)]
```

```r
coefsAllSel = coefsAll[paste0("GENE_",c('Ambra1', 'Ankfy1', 'Crebbp', 'Ep300', 'Fbxw7', 'Klhl6',
              'March6', 'Pparg', 'Rfwd2','Socs3', 'Syvn1', 'Traf2', 
              'Trip12', 'Wdr82'),"_"),extraGenesAvail]
rownames(coefsAllSel) = sapply(rownames(coefsAllSel),function(x){strsplit(x,"_")[[1]][2]})
```

```r
coef_comboSel = coef_combo[c('Ambra1', 'Ankfy1', 'Crebbp', 'Ep300', 'Fbxw7', 'Klhl6',
              'March6', 'Pparg', 'Rfwd2','Socs3', 'Syvn1', 'Traf2', 
              'Trip12', 'Wdr82'),extraGenesAvail]
```

```r
for(elem in rownames(coef_comboSel, coefsAllSel)){
        k = cor(unlist(coef_comboSel[elem,]), unlist(coefsAllSel[elem,]), method = "spearman")
        plot(unlist(coef_comboSel[elem,]),unlist(coefsAllSel[elem,]), main = paste0("cor = ",k))
}
```
