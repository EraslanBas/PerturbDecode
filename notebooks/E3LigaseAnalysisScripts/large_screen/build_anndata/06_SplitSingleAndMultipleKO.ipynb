# 06 · Split single- from multiple-knockout cells

A cell's knockout count is how many guides it carries. Cells with one guide or
none are the single-knockout population, which the guide QC and the effect-size
model are built on. Cells with several are the combinatorial population used in
the interaction models.

**Reads** `anndataFileName2`.
**Writes** `par_save_filename_5` and `par_save_filename_6`.

## Setup

```python
from libraries import *
from parameters import *

os.chdir(projectDir)
```

```python
adata = sc.read(anndataFileName2)
print(f"screen: {adata.shape[0]} cells x {adata.shape[1]} genes")

guide_names = list(adata.uns["feature_barcode_names"])
adata.obs["KONo"] = (adata.obs[guide_names] > 0).sum(axis=1).to_numpy()

print(adata.obs.KONo.value_counts().sort_index().to_string())
```

```python
adata_single = adata[adata.obs.KONo <= 1].copy()
adata_multiple = adata[adata.obs.KONo > 1].copy()

adata_single.write(par_save_filename_5)
print(f"single-knockout  : {adata_single.shape[0]} cells -> {par_save_filename_5}")

adata_multiple.write(par_save_filename_6)
print(f"multiple-knockout: {adata_multiple.shape[0]} cells -> {par_save_filename_6}")
```
